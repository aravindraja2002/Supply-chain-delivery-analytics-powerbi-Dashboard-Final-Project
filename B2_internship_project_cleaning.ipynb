```python
import pandas as pd
import numpy as np

df = pd.read_csv("dynamic_supply_chain_logistics_dataset.csv")
source_rows = df.shape[0]
print("Source rows:", source_rows, "| Columns:", df.shape[1])
df.head()
```

```python
df.info()
df.describe().T
```

```python
log = []
def add_log(step, issue, action, rows_after):
    log.append({"step": step, "issue": issue, "action": action, "rows_after": rows_after})

clean = df.copy()
add_log("Load", "Source file", "Loaded raw CSV", len(clean))
```

```python
clean.columns = clean.columns.str.strip().str.lower().str.replace(" ", "_")
clean["risk_classification"] = clean["risk_classification"].str.strip()

print(clean["risk_classification"].value_counts())
add_log("Standardize", "Column names and text spacing", "Trimmed and lowercased names; trimmed text", len(clean))
```

```python
clean["timestamp"] = pd.to_datetime(clean["timestamp"], errors="coerce")
print("Invalid dates:", clean["timestamp"].isna().sum())
clean["risk_classification"] = clean["risk_classification"].astype("category")

add_log("Types", "timestamp stored as text", "Converted to datetime", len(clean))
```

```python
print("Missing values per column:")
print(clean.isna().sum()[clean.isna().sum() > 0])
print("Total missing:", clean.isna().sum().sum())

dups = clean.duplicated().sum()
print("Duplicate rows:", dups)

clean = clean.drop_duplicates()
clean = clean.dropna(subset=["timestamp"])

add_log("Missing/duplicates", f"{dups} duplicates, 0 missing expected", "Dropped duplicates and rows with invalid dates", len(clean))
```

```python
score_cols = ["handling_equipment_availability", "order_fulfillment_status",
              "weather_condition_severity", "supplier_reliability_score",
              "cargo_condition_status", "driver_behavior_score",
              "fatigue_monitoring_score", "disruption_likelihood_score",
              "delay_probability"]
scale10_cols = ["traffic_congestion_level", "port_congestion_level", "route_risk_level"]

issues = {}
for c in score_cols:
    issues[c] = ((clean[c] < 0) | (clean[c] > 1)).sum()
for c in scale10_cols:
    issues[c] = ((clean[c] < 0) | (clean[c] > 10)).sum()
issues["lead_time_days"] = (clean["lead_time_days"] <= 0).sum()
issues["shipping_costs"] = (clean["shipping_costs"] <= 0).sum()
issues["warehouse_inventory_level"] = (clean["warehouse_inventory_level"] < 0).sum()
issues["historical_demand"] = (clean["historical_demand"] < 0).sum()

pd.Series(issues).sort_values(ascending=False)
```

```python
num_cols = ["lead_time_days", "shipping_costs", "warehouse_inventory_level",
            "fuel_consumption_rate", "customs_clearance_time",
            "loading_unloading_time", "delivery_time_deviation"]

rows = []
for c in num_cols:
    q1, q3 = clean[c].quantile([0.25, 0.75])
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n = ((clean[c] < low) | (clean[c] > high)).sum()
    rows.append({"column": c, "outliers": n, "pct": round(100 * n / len(clean), 2)})

pd.DataFrame(rows)
add_log("Outliers", "IQR check on numeric columns", "Reported only; kept all rows (bounded synthetic data)", len(clean))
```

```python
print("Negative eta_variation_hours:", (clean["eta_variation_hours"] < 0).sum())
print("Negative delivery_time_deviation:", (clean["delivery_time_deviation"] < 0).sum())

clean["early_arrival_flag"] = (clean["delivery_time_deviation"] < 0).astype(int)

add_log("Negative values", "Negative deviation values", "Kept; added early_arrival_flag (interpretation: possible early delivery, unconfirmed)", len(clean))
```

```python
clean["year"] = clean["timestamp"].dt.year
clean["month_num"] = clean["timestamp"].dt.month
clean["month"] = clean["timestamp"].dt.to_period("M").astype(str)
clean["hour"] = clean["timestamp"].dt.hour

clean["supplier_band"] = pd.cut(
    clean["supplier_reliability_score"],
    bins=[0, 0.33, 0.66, 1.0],
    labels=["Low", "Medium", "High"],
    include_lowest=True,
)

# Assumption: inventory below 50 counts as low inventory (a proxy, not a true stockout)
clean["low_inventory_flag"] = (clean["warehouse_inventory_level"] < 50).astype(int)

# Assumption: fulfillment score of 0.8 or higher counts as fulfilled
clean["fulfilled_flag"] = (clean["order_fulfillment_status"] >= 0.8).astype(int)

clean["high_risk_flag"] = (clean["risk_classification"] == "High Risk").astype(int)

add_log("Calculated fields", "Fields needed for dashboard", "Added year, month, hour, supplier_band, low_inventory_flag, fulfilled_flag, high_risk_flag", len(clean))
```

```python
print("Source rows:", source_rows)
print("Cleaned rows:", len(clean))
print("Rows removed:", source_rows - len(clean))
print("Remaining missing values:", clean.isna().sum().sum())

pd.DataFrame(log)
```

```python
kpis = {
    "Average lead time (days)": clean["lead_time_days"].mean(),
    "Average shipping cost": clean["shipping_costs"].mean(),
    "High-risk share (%)": clean["high_risk_flag"].mean() * 100,
    "Low-inventory frequency (%)": clean["low_inventory_flag"].mean() * 100,
    "Fulfilled share (%)": clean["fulfilled_flag"].mean() * 100,
    "Average delay probability": clean["delay_probability"].mean(),
}
kpi_table = pd.DataFrame(list(kpis.items()), columns=["KPI", "Python value"]).round(2)
kpi_table
```

```python
clean.to_csv("B2_prepared.csv", index=False)
pd.DataFrame(log).to_csv("B2_cleaning_log.csv", index=False)
kpi_table.to_csv("B2_kpi_python_values.csv", index=False)

from google.colab import files
files.download("B2_prepared.csv")
files.download("B2_cleaning_log.csv")
files.download("B2_kpi_python_values.csv")
```
